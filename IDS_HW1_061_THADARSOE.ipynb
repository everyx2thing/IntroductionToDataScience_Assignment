### 2-4. Groupby 집계 (2pt)

범주형 변수 1개를 기준으로 수치형 변수를 집계(`groupby`)하세요.

````python
# groupby 집계
# 예: df.groupby('Team')['Score'].mean()

# 직종(job_role)별 평균 자동화 위험도 집계
groupby_result = df.groupby('job_role')['automation_risk_percent'].mean().sort_values(ascending=False)
print(groupby_result)
````

**해석:** 직종별 평균 자동화 위험도를 집계한 결과 Truck Driver(약 60.7%)와 Customer Support Rep(약 59.9%)가 가장 높은 자동화 위험도를 보였다.
반면 Software Engineer(약 34.9%)와 Data Analyst(약 35.5%)는 가장 낮은 위험도를 기록했다.
반복적이고 규칙적인 업무를 수행하는 직종일수록 자동화 위험이 높고 창의적이거나 기술적 역량이 요구되는 직종은 상대적으로 낮다는 것을 알 수 있다.

### 2-5. 흥미로운 발견 (2pt)

데이터를 탐색하면서 발견한 **흥미로운 패턴이나 특이점**을 1가지 이상 서술하세요.

코드를 통해 근거를 제시하세요.

````python
# 흥미로운 패턴을 발견하기 위한 코드
# 예: 정렬, 필터링, 상관관계 등

salary_by_risk = df.groupby('automation_risk_category')['salary_change_percent'].mean().round(3)
print(salary_by_risk)


corr = df['automation_risk_percent'].corr(df['salary_change_percent'])
print(f"\n상관계수: {corr:.4f}")
````

**발견:** 자동화 위험도가 높을수록 임금이 더 많이 감소할 것이라고 예상했지만 실제 데이터는 그렇지 않았다.
High 위험군의 평균 임금 변화율(+0.076%)과 Low 위험군(+0.220%)의 차이가 거의 없으며 `automation_risk_percent`와 `salary_change_percent`의 상관계수도 -0.005로 사실상 0에 가깝다.
이는 자동화 위험도 단독으로는 임금 변화를 설명하지 못하고 훨씬 더 복합적인 요인에 의해 결정된다는 점을 얘기하는 발견이다.

### 3-1. 분포 확인 — 히스토그램 또는 박스플롯

## Task 4. 가설 설정 및 검증 (25pt)

데이터를 바탕으로 **가설(Hypothesis)을 1개 설정**하고, 분석 및 시각화를 통해 검증하세요.

### 4-1. 가설 설정 (6pt)

구체적인 주장 1개를 작성하세요.

**가설:** (예: "나이가 많을수록 구매 금액이 높을 것이다.")

> 국가별 평균 AI 도입 수준이 높을수록 평균 급여 변화율도 높을 것이다.

### 4-2. 분석 및 시각화 (7pt)

가설을 검증하기 위한 분석을 수행하세요. 관련 시각화를 1개 이상 포함하세요.

````python
# 가설 검증을 위한 분석 코드

# 여기에 코드를 작성하세요
from scipy import stats

country_stats = df.groupby('country').agg(
    avg_ai_adoption=('ai_adoption_level', 'mean'),
    avg_salary_change=('salary_change_percent', 'mean')
).reset_index()

r, p = stats.pearsonr(country_stats['avg_ai_adoption'], country_stats['avg_salary_change'])
print(f"Pearson r = {r:.4f}, p-value = {p:.4f}")
display(country_stats)
````

````python
# 가설 관련 시각화
plt.figure(figsize=(8, 5))

# 여기에 코드를 작성하세요
plt.scatter(country_stats['avg_ai_adoption'], country_stats['avg_salary_change'], color='steelblue', s=100)

for _, row in country_stats.iterrows():
    plt.annotate(row['country'], (row['avg_ai_adoption'], row['avg_salary_change']), textcoords="offset points", xytext=(5,5), fontsize=9)

plt.title(f'국가별 AI 도입 수준과 급여 변화율 관계 (r={r:.4f}, p={p:.4f})')
plt.xlabel('평균 AI 도입 수준')
plt.ylabel('평균 급여 변화율 (%)')
plt.axhline(0, color='gray', linestyle='--')
plt.tight_layout()
plt.show()
````

### 4-3. 결론 (6pt)

가설이 지지되는지 여부를 판단하고, 그 근거를 서술하세요.

**결론:** 가설은 명확하게 지지된다고 보기 어렵다. 분석 결과, AI 도입 수준과 임금 변화율 사이에는 r = 0.60으로 중간 정도의 양의 상관관계가 나타났다. 이는 AI 도입 수준이 높을수록 임금 변화율이 증가하는 경향이 있음을 시사한다. 그러나 p-value = 0.088로 유의수준 0.05를 초과하여, 통계적으로 유의하다고 보기 어렵다.

따라서 본 데이터에서는 가설을 뒷받침하는 충분한 근거가 있다고 결론 내리기 어렵다.

### 4-4. 인과관계 vs 상관관계 검토 (6pt)

수업 Topic 5 (Causality)에서 배운 내용을 적용하여 다음에 답하세요:

- 위 분석 결과가 **인과관계(causation)**를 보여주는가, 아니면 **상관관계(association)**에 불과한가?
- 이 결과에 영향을 줄 수 있는 **교란변수(confounding variable)**가 있다면 무엇인가?
- 인과관계를 확인하려면 어떤 실험/추가 분석이 필요한가?

**답변:** 이 분석 결과는 Association에 불과하다. Pearson 상관계수는 두 변수 간의 선형적 관련성만을 측정하며 한 변수가 다른 변수를 실제로 Cause하는지는 알 수 없다. 또한 p-value = 0.088로 통계적으로 유의하지 않기 때문에 신뢰할 수 있는 인과관계를 주장하기에는 한계가 있다.

교란변수(Confounding Variable)
- 경제 발전 수준: 선진국일수록 AI 도입도 높고 임금 구조도 다를 수 있음  
- 노동 시장 제도: 최저임금, 노동법 등 정책 요인이 임금 변화에 영향

인과관계 확인을 위해 필요한 분석
- Natural Experiment: 정책 변화 전후 비교  
- 패널 데이터 분석: 연도별 변화 흐름 분석  
- 회귀분석: GDP, 산업 구조 등 여러 변수 포함
